## Images loading

This function is used to convert an image into a matrix using a path specified in the parameters. This will enable the machine learning program to interpret the data contained in the image.

````python
def convert_img(img_path):
    matrice_img = plt.imread(img_path)
    return matrice_img
````

This function is used to make an array of matrice for specific folder

````python
def makeTabImg(path):
    dossier = path
    tab = []
    for file in os.listdir(dossier):
        matrice_img = convert_img(path + "/" + file)
        tab.append(matrice_img)
    return tab
````
