```python
import os
import sys
import numpy as np
from PIL import Image

def analyser_dimensions(chemin_dossier):
    if not os.path.exists(chemin_dossier):
        sys.stderr.write(f"Erreur : Le dossier '{chemin_dossier}' n'existe pas.\n")
        sys.exit(84)

    largeurs = []
    hauteurs = []

    for nom_fichier in os.listdir(chemin_dossier):
        if nom_fichier.lower().endswith(('.png', '.jpg', '.jpeg')):
            chemin_complet = os.path.join(chemin_dossier, nom_fichier)
            try:
                with Image.open(chemin_complet) as img:
                    largeur, hauteur = img.size
                    largeurs.append(largeur)
                    hauteurs.append(hauteur)
            except Exception as e:
                sys.stderr.write(f"Erreur avec {nom_fichier}: {e}\n")
                sys.exit(84)

    if not largeurs:
        print(f"Aucune image trouvée dans {chemin_dossier}.")
        sys.exit(0)

    largeur_moy = int(np.mean(largeurs))
    hauteur_moy = int(np.mean(hauteurs))
    
    largeur_min = np.min(largeurs)
    hauteur_min = np.min(hauteurs)
    
    largeur_max = np.max(largeurs)
    hauteur_max = np.max(hauteurs)

    print(f"--- Analyse du dossier : {chemin_dossier} ---")
    print(f"Nombre d'images    : {len(largeurs)}")
    print(f"Taille moyenne     : {largeur_moy}x{hauteur_moy}")
    print(f"Taille la plus petite : {largeur_min}x{hauteur_min}")
    print(f"Taille la plus grande : {largeur_max}x{hauteur_max}")

if __name__ == "__main__":
    dossier_cible = "chest_xray/test/NORMAL"
    analyser_dimensions(dossier_cible)
    dossier_cible = "chest_xray/test/PNEUMONIA"
    analyser_dimensions(dossier_cible)
    dossier_cible = "chest_xray/train/NORMAL"
    analyser_dimensions(dossier_cible)
    dossier_cible = "chest_xray/train/PNEUMONIA"
    analyser_dimensions(dossier_cible)
    sys.exit(0)
```
