# normalisation du dataset

## les imports

```python
import os
import numpy as np
import matplotlib.pyplot as plt
```

## la fonction de normalisation du dataset

```python
def load_data(base_path):
    label_map = {'normal': 0, 'bacteria': 1, 'virus': 2}   
    X = []
    y = []
    for category, label_value in label_map.items():
        category_path = os.path.join(base_path, category)
        if not os.path.exists(category_path):
            continue
        for file_name in os.listdir(category_path):
            file_path = os.path.join(category_path, file_name)
            try:
                img = plt.imread(file_path)
                if img.max() > 1.0:
                    img = img / 255.0
                img_flattened = img.flatten()
                X.append(img_flattened)
                y.append(label_value)
            except Exception as e:
                print(f"Erreur avec l'image {file_path}: {e}")
    return np.array(X), np.array(y)

print("Chargement des données d'entraînement...")
X_train, y_train = load_data('dataset/train')
print("Chargement des données de test...")
X_test, y_test = load_data('dataset/test')
print(f"Dimensions des données d'entraînement : {X_train.shape}")
print(f"Dimensions des données de test : {X_test.shape}")
```
