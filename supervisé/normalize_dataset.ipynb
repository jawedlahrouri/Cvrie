# normalisation du dataset

## les imports

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.neighbors import KNeighborsClassifier
```

## la fonction de normalisation du dataset

```python
def load_data(base_path):
    label_map = {'normal': 0, 'bacteria': 1, 'virus': 2}   
    X = []
    y = []
    for category, label_value in label_map.items():
        category_path = os.path.join(base_path, category)
        if not os.path.exists(category_path):
            continue
        for file_name in os.listdir(category_path):
            file_path = os.path.join(category_path, file_name)
            try:
                img = plt.imread(file_path)
                if img.max() > 1.0:
                    img = img / 255.0
                img_flattened = img.flatten()
                X.append(img_flattened)
                y.append(label_value)
            except Exception as e:
                print(f"Erreur avec l'image {file_path}: {e}")
    return np.array(X), np.array(y)

print("Chargement des données d'entraînement...")
X_train, y_train = load_data('dataset/train')
print("Chargement des données de test...")
X_test, y_test = load_data('dataset/test')
print(f"Dimensions des données d'entraînement : {X_train.shape}")
print(f"Dimensions des données de test : {X_test.shape}")
```

## Entraînement et évaluation du modèle (Random Forest)
Nous allons tester un algorithme de Random Forest sur nos données aplaties.

```python
print("Initialisation du modèle Random Forest...")
# n_jobs=-1 permet d'utiliser tous les coeurs de ton processeur
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)

print("Entraînement en cours (patience, il y a 65 536 colonnes à analyser...)...")
rf_model.fit(X_train, y_train)
print("Entraînement terminé !")

print("Prédiction sur le jeu de test...")
y_pred = rf_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"\nPrécision globale (Accuracy) : {accuracy * 100:.2f}%")

print("\nRapport de classification détaillé :")
print(classification_report(y_test, y_pred, target_names=['Normal', 'Bacteria', 'Virus']))
```

```python
print("Initialisation du modèle KNN (K=3)...")
knn_model = KNeighborsClassifier(n_neighbors=3, n_jobs=-1)

print("Entraînement du KNN en cours...")
knn_model.fit(X_train, y_train)
print("Entraînement terminé !")

print("Prédiction sur le jeu de test (cela peut prendre du temps)...")
y_pred_knn = knn_model.predict(X_test)

accuracy_knn = accuracy_score(y_test, y_pred_knn)
print(f"\nPrécision globale KNN (Accuracy) : {accuracy_knn * 100:.2f}%")

print("\nRapport de classification détaillé (KNN) :")
print(classification_report(y_test, y_pred_knn, target_names=['Normal', 'Bacteria', 'Virus']))
```
