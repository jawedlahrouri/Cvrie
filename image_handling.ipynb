## Chargement des images

Cette fonction sert a convertir une image en matrice grace a un path donné en parametres. Cela vas servir a faire en sorte que le programme de machine learning puisse interpreter les données contenues dans l'image.

````python
def convert_img(img_path):
    matrice_img = plt.imread(img_path)
    return matrice_img
````

Cette fonction a pour but de generer un tableau de matrice afin de pouvoir donner des donner facile a utiliser au programe de machine learning en quatité.

````python
def makeTabImg(path):
    dossier = path
    tab = []
    for file in os.listdir(dossier):
        matrice_img = convert_img(path + "/" + file)
        tab.append(matrice_img)
    return tab
````
